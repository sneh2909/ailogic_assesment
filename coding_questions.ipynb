# Find Minimum in Rotated Sorted Array

Input:  nums = [3, 4, 5, 1, 2]
Output: 1
 
Input:  nums = [4, 5, 6, 7, 0, 1, 2]
Output: 0



```python
def min_num(nums):
    left=0
    right=len(nums)-1
    while left < right:
        mid=(left+right)//2
        if nums[mid]>nums[right]:
            left=mid+1
        else:
            right=mid
    return nums[left]
```

```python
nums=[1,2,3,4,5] # as list is already sorted
min_num(nums)
```

# Longest Substring Without Repeating Characters

Input:  s = "abcabcbb"
Output: 3   (substring "abc")
 
Input:  s = "bbbbb"
Output: 1   (substring "b")



```python
def longest_substring(s):
    seen={}
    left=0
    max_length=0
    for i,x in enumerate(s):
        if x in seen and seen[x]>=left:
            left=seen[x]+1
        seen[x]=i
        length=i-left+1
        max_length=max(length,max_length)
    return max_length
```

```python
s = "bbbbb"
longest_substring(s)
```

```python
s = "abcabcbb"
longest_substring(s)
```
